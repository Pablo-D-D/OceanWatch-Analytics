## Cruce de las capas para la generación del área que permite identificar los puertos de interés

### Concatenación de las geometrías en un mismo dataframe

```python
import geopandas as gpd
import pandas as pd

# Seleccionar únicamente las geometrías
terrestre = gdf_terrestre_wgs84[["geometry"]].copy()
maritima = gdf_maritima_filtrada[["geometry"]].copy()

# Identificar el origen de cada geometría
terrestre["origen"] = "Terrestre"
maritima["origen"] = "Marítimo"

# Integrar ambas capas
gdf_integrada = gpd.GeoDataFrame(
    pd.concat(
        [terrestre, maritima],
        ignore_index=True
    ),
    geometry="geometry",
    crs="EPSG:4326"
)

# Verificar resultados
print("RESULTADOS DE LA INTEGRACIÓN")
print("-" * 50)

print(f"Registros terrestres: {len(terrestre)}")
print(f"Registros marítimos: {len(maritima)}")
print(f"Registros integrados: {len(gdf_integrada)}")

print(f"\nCRS: {gdf_integrada.crs}")

print("\nTIPOS DE GEOMETRÍA")
print(gdf_integrada.geom_type.value_counts())

print("\nREGISTROS POR ORIGEN")
print(gdf_integrada["origen"].value_counts())

print("\nGEOMETRÍAS VÁLIDAS")
print(gdf_integrada.is_valid.value_counts())
```

### Dissolve de las geometrías para la generación de una capa mas sencilla

```python

# Dissolve general de todas las geometrías
gdf_limites_final = gdf_integrada.dissolve()

# Reiniciar índice
gdf_limites_final = gdf_limites_final.reset_index(drop=True)

# Verificar resultados
print("RESULTADOS DEL DISSOLVE")
print("-" * 50)

print(f"Registros originales: {len(gdf_integrada)}")
print(f"Registros después del dissolve: {len(gdf_limites_final)}")

print(f"\nCRS: {gdf_limites_final.crs}")

print("\nTIPO DE GEOMETRÍA")
print(gdf_limites_final.geom_type.value_counts())

print("\nVALIDEZ GEOMÉTRICA")
print(gdf_limites_final.is_valid.value_counts())

print("\nTOTAL DE PARTES GEOMÉTRICAS")
geom = gdf_limites_final.geometry.iloc[0]

if geom.geom_type == "MultiPolygon":
    print(len(geom.geoms))
else:
    print(1)
```

### Visualización de la capa integrada

```python

import folium

# Crear mapa base
mapa_limites_final = folium.Map(
    location=[39.5, -98.35],
    zoom_start=3,
    tiles="OpenStreetMap"
)

# Agregar geometría disuelta
folium.GeoJson(
    gdf_limites_final,
    name="Límites terrestres y marítimos US",
    style_function=lambda feature: {
        "fillColor": "#8DB580",
        "color": "#3D6B35",
        "weight": 1.5,
        "fillOpacity": 0.5
    }
).add_to(mapa_limites_final)

# Control de capas
folium.LayerControl().add_to(mapa_limites_final)

# Visualizar mapa
displayHTML(mapa_limites_final._repr_html_())
```

Esta capa será la finalmente nos permita hacer la selección de los puertos que se encuentran al interior de los límites terrestres y marítimos de estados unidos

## Almacenamiento de la capa final de límites

```python
import os
import shutil
import zipfile
import tempfile
import geopandas as gpd

# Configuración de salida
directorio_salida = (
    "/Volumes/workspace/oceanwatch/geo_data/Processed/"
)

nombre_capa = "Limites_Terrestres_Maritimos_US_WGS84"

ruta_zip_final = os.path.join(
    directorio_salida,
    f"{nombre_capa}.zip"
)

# Crear directorio temporal automáticamente
with tempfile.TemporaryDirectory() as directorio_temporal:

    # Ruta del Shapefile temporal
    ruta_shape = os.path.join(
        directorio_temporal,
        f"{nombre_capa}.shp"
    )

    # Exportar Shapefile
    gdf_limites_final.to_file(
        ruta_shape,
        driver="ESRI Shapefile",
        encoding="UTF-8"
    )

    # Ruta del ZIP temporal
    ruta_zip_temporal = os.path.join(
        directorio_temporal,
        f"{nombre_capa}.zip"
    )

    # Comprimir archivos del Shapefile
    with zipfile.ZipFile(
        ruta_zip_temporal,
        "w",
        zipfile.ZIP_DEFLATED
    ) as zipf:

        for archivo in os.listdir(directorio_temporal):
            if archivo.endswith((
                ".shp", ".shx", ".dbf",
                ".prj", ".cpg"
            )):
                zipf.write(
                    os.path.join(directorio_temporal, archivo),
                    arcname=archivo
                )

    # Copiar ZIP a Unity Catalog
    os.makedirs(directorio_salida, exist_ok=True)

    shutil.copyfile(
        ruta_zip_temporal,
        ruta_zip_final
    )

# Verificar archivo exportado
print("EXPORTACIÓN COMPLETADA")
print("-" * 50)
print(f"Archivo: {ruta_zip_final}")
print(
    f"Tamaño: {os.path.getsize(ruta_zip_final) / (1024**2):.2f} MB"
)

```

## Verificación del archivo generado

```python

# Leer capa exportada
gdf_verificacion = gpd.read_file(ruta_zip_final)

# Verificaciones
print("VALIDACIÓN DEL ARCHIVO PROCESADO")
print("-" * 50)

print(f"Registros: {len(gdf_verificacion)}")
print(f"CRS: {gdf_verificacion.crs}")
print(f"Tipo de geometría: {gdf_verificacion.geom_type.iloc[0]}")
print(f"Geometría válida: {gdf_verificacion.is_valid.all()}")

# Contar partes geométricas
geom = gdf_verificacion.geometry.iloc[0]

partes = (
    len(geom.geoms)
    if geom.geom_type == "MultiPolygon"
    else 1
)

print(f"Partes geométricas: {partes}")

# Comparar geometría original y exportada
print(
    "Geometría conservada:",
    gdf_limites_final.geometry.iloc[0].equals(geom)
)
```
